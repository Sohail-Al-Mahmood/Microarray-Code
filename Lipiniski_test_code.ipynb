```python
def lipinski_test(smiles):
    mol = Chem.MolFromSmiles(smiles)
    if mol is None:
        return "Invalid SMILES"
    
    mw = Descriptors.MolWt(mol)
    logp = Descriptors.MolLogP(mol)
    h_donor = Descriptors.NumHDonors(mol)
    h_acceptor = Descriptors.NumHAcceptors(mol)
    
    violations = 0
    if mw > 500: violations += 1
    if logp > 5: violations += 1
    if h_donor > 5: violations += 1
    if h_acceptor > 10: violations += 1
    
    return {
        "MW": mw,
        "LogP": logp,
        "H-Donors": h_donor,
        "H-Acceptors": h_acceptor,
        "Violations": violations,
        "Passed": violations <= 1
    }
```

```python
# 🔍 Test on all drugs
for name, smiles in drug_data.items():
    result = lipinski_test(smiles)
    print(f"\n{name}")
    for k, v in result.items():
        print(f"  {k}: {v}")
```

```python
# Assuming you already have drug_data and lipinski_test() defined
# To save the output: 

passed = []
failed = []

with open("lipinski_results.txt", "w", encoding="utf-8") as file:
    for name, smiles in drug_data.items():
        result = lipinski_test(smiles)
        file.write(f"{name}\n")
        for key, value in result.items():
            file.write(f"  {key}: {value}\n")
        file.write("\n")

        if result.get("Passed"):
            passed.append(name)
        else:
            failed.append(name)

    # Summary section
    file.write("==== Summary ====\n")
    file.write(f"✅ Total Passed (True): {len(passed)}\n")
    for drug in passed:
        file.write(f"  - {drug}\n")

    file.write(f"\n❌ Total Failed (False): {len(failed)}\n")
    for drug in failed:
        file.write(f"  - {drug}\n")

print("✅ Lipinski test results saved to 'lipinski_results.txt'")
```
