```python
import pubchempy as pcp
```

```python
# 🔠 List of drug names
drug_names = [
    "Irinotecan", "Midostaurin", "Drospirenone", "Netupitant", "Acalabrutinib",
    "Tucatinib", "Tepotinib", "Imatinib", "Trilaciclib", "Alectinib",
    "Bexarotene", "Epirubicin", "Ibrutinib", "Cabozantinib", "Tazemetostat",
    "Idarubicin", "Abemaciclib", "Abiraterone", "Etoposide", "Doxorubicin",
    "Rebamipide", "Testolactone", "Progesterone", "Tivozanib", "Estrone sulfate",
    "Vimseltinib", "Ecamsule", "Axitinib", "Gilteritinib", "Lazertinib",
    "Tirbanibulin", "Selpercatinib", "Pyrvinium", "Raloxifene", "Aprepitant"
]
```

```python
len(drug_names)
```

```python
# 📦 List to store results
results = []

for name in drug_names:
    compounds = pcp.get_compounds(name, 'name')
    if compounds:
        smiles = compounds[0].canonical_smiles
        results.append(f"{name}\t{smiles}")
        print(f"{name} → {smiles}")
    else:
        results.append(f"{name}\tNOT FOUND")
        print(f"{name} → Not found on PubChem")
```

```python
# 💾 Save the results to a text file
output_path = r"C:/Users/ASUS/Desktop/my python/drug name to smiles code/drug_smiles.txt"
with open(output_path, "w") as file:
    for line in results:
        file.write(line + "\n")

print(f"\n✅ All SMILES saved to: {output_path}")
```
